# GDAL Format Conversion Notebook
Template for converting between geospatial formats.

```python
import subprocess

# GeoJSON -> GeoPackage
subprocess.run(['ogr2ogr', '-f', 'GPKG', 'output.gpkg', 'input.geojson'], check=True)

# GeoTIFF -> COG
subprocess.run(['gdal_translate', '-of', 'COG', 'input.tif', 'output_cog.tif'], check=True)
print('Conversions complete')
```
