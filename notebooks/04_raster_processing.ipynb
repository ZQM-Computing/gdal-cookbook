# GDAL Raster Processing Notebook
Template for raster analysis with GDAL.

```python
from osgeo import gdal
import numpy as np

ds = gdal.Open('input.tif')
band = ds.GetRasterBand(1)
arr = band.ReadAsArray()
print('Shape:', arr.shape)
print('Min/Max:', arr.min(), arr.max())
```

```python
# Compute NDVI
nir = gdal.Open('nir.tif').GetRasterBand(1).ReadAsArray().astype(float)
red = gdal.Open('red.tif').GetRasterBand(1).ReadAsArray().astype(float)
ndvi = (nir - red) / (nir + red + 1e-10)
print('NDVI range:', ndvi.min(), ndvi.max())
```
