# GDAL Python API Notebook
Template for using osgeo Python bindings directly.

```python
from osgeo import gdal, ogr, osr
import numpy as np

# Create in-memory raster
driver = gdal.GetDriverByName('MEM')
ds = driver.Create('', 128, 128, 1, gdal.GDT_Byte)
srs = osr.SpatialReference(); srs.ImportFromEPSG(4326)
ds.SetProjection(srs.ExportToWkt())
ds.SetGeoTransform([0, 1, 0, 0, 0, -1])
arr = np.tile(np.arange(128, dtype=np.uint8), (128, 1))
ds.GetRasterBand(1).WriteArray(arr)
print('Raster created:', ds.RasterXSize, 'x', ds.RasterYSize)
```

```python
# Geometry operations
geom = ogr.CreateGeometryFromWkt('POINT(-73.9857 40.7484)')
buffered = geom.Buffer(0.01)
print('Buffered area:', buffered.GetArea())
print(buffered.ExportToWkt()[:100])
```
