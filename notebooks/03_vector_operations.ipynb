# GDAL Vector Operations Notebook
Template for vector data processing with OGR.

```python
from osgeo import ogr, osr
import subprocess

# Open a vector dataset
ds = ogr.Open('input.shp')
layer = ds.GetLayer(0)
print('Features:', layer.GetFeatureCount())
print('Geometry type:', layer.GetGeomType())
```

```python
# Reproject vector
subprocess.run(['ogr2ogr', '-t_srs', 'EPSG:4326', 'output.shp', 'input.shp'], check=True)
print('Reprojected to EPSG:4326')
```
